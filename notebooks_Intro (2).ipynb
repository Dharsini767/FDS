```python
import numpy as np
import matplotlib.pyplot as plt

population_mean = 50
population_std = 10
population_size = 100000

population = np.random.normal(
    population_mean,
    population_std,
    population_size
)

sample_sizes = [30, 50, 100]
num_samples = 1000

sample_means = {}

for size in sample_sizes:
    sample_means[size] = []

    for _ in range(num_samples):
        sample = np.random.choice(
            population,
            size=size,
            replace=False
        )

        sample_means[size].append(np.mean(sample))

plt.figure(figsize=(12, 8))

for i, size in enumerate(sample_sizes):
    plt.subplot(len(sample_sizes), 1, i + 1)

    plt.hist(
        sample_means[size],
        bins=30,
        alpha=0.7,
        label=f'Sample Size {size}'
    )

    plt.axvline(
        np.mean(population),
        color='red',
        linestyle='dashed',
        linewidth=1.5,
        label='Population Mean'
    )

    plt.title(f'Sampling Distribution (Sample Size {size})')
    plt.xlabel('Sample Mean')
    plt.ylabel('Frequency')
    plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import scipy.stats as stats

sample_data = np.array([
    152, 148, 151, 149, 147, 153, 150, 148, 152, 149,
    151, 150, 149, 152, 151, 148, 150, 152, 149, 150,
    148, 153, 151, 150, 149, 152, 148, 151, 150, 153
])

population_mean = 150

sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)

n = len(sample_data)

z_statistic = (sample_mean - population_mean) / (sample_std / np.sqrt(n))

p_value = 2 * (1 - stats.norm.cdf(np.abs(z_statistic)))

print(f"Sample Mean: {sample_mean:.2f}")
print(f"Z-Statistic: {z_statistic:.4f}")
print(f"P-Value: {p_value:.4f}")

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis: The average weight is significantly different from 150 grams.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference in average weight from 150 grams.")
```

```python
def outDetection(array):
    sorted(array)
    Q1, Q3 = np.percentile(array, [25, 75])
    IQR = Q3 - Q1
    lr = Q1 - (1.5 * IQR)
    ur = Q3 + (1.5 * IQR)
    return lr, ur

lr, ur = outDetection(array)
lr, ur
```

```python
import seaborn as sns
%matplotlib inline
sns.displot(array)
```

```python
sns.distplot(array)
```

```python
new_array=array[(array>lr) & (array<ur)]
new_array
```

```python
sns.displot(new_array)
```

```python
lr1,ur1=outDetection(new_array)
lr1,ur1
```

```python
final_array=new_array[(new_array>lr1) & (new_array<ur1)]
final_array
```

```python
sns.distplot(final_array)
```

```python
import pandas as pd 
 
import matplotlib.pyplot as plt 
 
from statsmodels.tsa.seasonal import seasonal_decompose 
data = pd.read_csv('Daily Website Traffic.csv', parse_dates=['Day Index'])
data = data.set_index('Day Index').sort_index()
decomposition = seasonal_decompose(
    data['Visits'], model='additive', period=7
)
decomposition.plot()
plt.show()
```

```python
data['Visits_MA'] = data['Visits'].rolling(window=30).mean()
data[['Visits', 'Visits_MA']].plot()
plt.title('Website Visits with Moving Average')
plt.xlabel('Date')
plt.ylabel('Visits')
plt.show()
```

```python
data['Visits_Z'] = (
    (data['Visits'] - data['Visits'].mean())
    / data['Visits'].std()
)

anomalies = data[data['Visits_Z'].abs() > 3]

plt.figure(figsize=(10, 5))
plt.plot(data.index, data['Visits'], label='Visits')
plt.scatter(
    anomalies.index,
    anomalies['Visits'],
    color='red',
    label='Anomalies'
)
plt.title('Website Traffic Anomaly Detection')
plt.xlabel('Date')
plt.ylabel('Visits')
plt.legend() 
plt.show()
```
